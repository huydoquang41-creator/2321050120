```python
import pandas as pd
path = 'Data_Patient.csv'
df_patient = pd.read_csv(path,index_col = 0)
df_patient.info()
print('\nDữ liệu 10 dòng đầu:\n', df_patient.head(10))
print('\nDữ liệu 5 dòng cuối:\n', df_patient.tail(5))
df_patient.rename(columns = {'feature_1':'Age',
                             'feature_2':'Gender',
                             'feature_3':'Type',
                             'feature_4':'Blood_pressure',
                             'feature_5':'Cholesterol',
                             'feature_6':'Heartbeat',
                             'feature_7':'Thalassemia',
                             'feature_8':'Result'},inplace = True)
df_patient.info()
```

Yêu cầu 3.2

```python
tk1=df_patient[['Age','Cholesterol']].describe()
print(tk1)
tk2=df_patient[['Gender','Type']].describe()
print('\n',tk2)
```

Yêu cầu 3.3

```python
colnull = df_patient.columns[df_patient.isnull().sum() > 0].tolist()
print('Các cột chứa gia trị null:\n',colnull)
print('Số lượng missing data:')
print(df_patient.isnull().sum())
```

Yêu cầu 3.4

```python
tt1 = df_patient.loc[['Patient_100','Patient_150','Patient_200'],:]
print('Thông tin bệnh nhân 100,150,200:')
print(tt1)
tt2 = df_patient.loc[['Patient_255','Patient_256','Patient_257','Patient_258','Patient_259','Patient_260'],['Age','Gender','Result']]
print('Thông tin tuổi,giới tính và kết của của bệnh nhân 255-260:')
print(tt2)
```

THỰC HÀNH 4

Yêu cầu 4.1

```python
path = 'Data_Patient.csv'
df_patient = pd.read_csv(path,index_col = 0)
df_patient.rename(columns = {'feature_1':'Age',
                             'feature_2':'Gender',
                             'feature_3':'Type',
                             'feature_4':'Blood_pressure',
                             'feature_5':'Cholesterol',
                             'feature_6':'Heartbeat',
                             'feature_7':'Thalassemia',
                             'feature_8':'Result'},inplace = True)
df_patient.head()
```

Yêu cầu 4.2

```python
df_male = df_patient[df_patient['Gender']=='Male']
df_female = df_patient[df_patient['Gender']=='Female']
df_no = df_patient[df_patient['Result']==0]
df_yes = df_patient[df_patient['Result']==1]
print('đàn ông:\n',df_male.head())
print('Phụ nữ:\n',df_female.head())
print('no\n',df_no.head())
print('yes\n',df_yes.head())
```

Yêu cầu 4.3

```python
df1 = df_patient[(df_patient['Result'] == 1) & (df_patient['Age'] >= 70)]
df2 = df_patient[(df_patient['Gender'] == 'Female') & (df_patient['Blood_pressure'] >= 170) & (df_patient['Result'] == 0)]
df3 = df_patient[(df_patient['Type'] == 'Typical angina') & (df_patient['Gender'] == 'Male') & (df_patient['Result'] == 1)]
print('Lọc 1:\n',df1)
print('Lọc 2:\n',df2)
print('Lọc 3:\n',df3)
```

Yêu cầu 4.4

```python
print('1. Chỉ số huyết áp:')
print('Thấp nhất:',df_patient['Blood_pressure'].min())
print('Cao nhất:',df_patient['Blood_pressure'].max())
print('Trung bình:',df_patient['Blood_pressure'].mean())
print('Trung vị:',df_patient['Blood_pressure'].median())
print('Độ lệch chuẩn:',df_patient['Blood_pressure'].std())
print('2. Chỉ số nhịp tim:')
print('Thấp nhất:',df_patient['Heartbeat'].min())
print('Cao nhất:',df_patient['Heartbeat'].max())
print('Trung bình:',df_patient['Heartbeat'].mean())
print('Trung vị:',df_patient['Heartbeat'].median())
print('Độ lệch chuẩn:',df_patient['Heartbeat'].std())
```
